```python
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator

# PARAMETERS (LOCK THESE)
IMG_SIZE = 224
BATCH_SIZE = 32
EPOCHS = 15

train_dir = "Warp-C/train"
val_dir = "Warp-C/val"

# DATA GENERATORS
train_gen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=20,
    zoom_range=0.2,
    horizontal_flip=True
)

val_gen = ImageDataGenerator(rescale=1./255)

train_data = train_gen.flow_from_directory(
    train_dir,
    target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    class_mode='categorical'
)

val_data = val_gen.flow_from_directory(
    val_dir,
    target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    class_mode='categorical'
)

print("CLASS INDICES:", train_data.class_indices)

# MODEL
base_model = tf.keras.applications.MobileNetV2(
    input_shape=(IMG_SIZE, IMG_SIZE, 3),
    include_top=False,
    weights="imagenet"
)

base_model.trainable = False  # IMPORTANT

model = tf.keras.Sequential([
    base_model,
    tf.keras.layers.GlobalAveragePooling2D(),
    tf.keras.layers.Dense(128, activation='relu'),
    tf.keras.layers.Dropout(0.3),
    tf.keras.layers.Dense(train_data.num_classes, activation='softmax')
])

model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

# TRAINING
history = model.fit(
    train_data,
    validation_data=val_data,
    epochs=EPOCHS
)
```

```python
model.save("warp_waste_classifier.h5")
```

```python
import tensorflow as tf
import numpy as np
import os
from tensorflow.keras.preprocessing import image
import pandas as pd

# Load trained model
model = tf.keras.models.load_model("warp_waste_classifier.h5")

IMG_SIZE = 224

# ⚠️ ONLY CLASSES USED IN TRAINING
labels = ['metal', 'paper', 'plastic']  # MUST match class_indices order
valid_classes = ['metal', 'paper', 'plastic']

results = []

base_dir = "Warp-C/val"

for true_class in valid_classes:
    class_dir = os.path.join(base_dir, true_class)
    if not os.path.isdir(class_dir):
        continue

    for img_name in os.listdir(class_dir):
        img_path = os.path.join(class_dir, img_name)

        img = image.load_img(img_path, target_size=(IMG_SIZE, IMG_SIZE))
        img_array = image.img_to_array(img) / 255.0
        img_array = np.expand_dims(img_array, axis=0)

        pred = model.predict(img_array, verbose=0)
        pred_index = np.argmax(pred)
        pred_class = labels[pred_index]
        confidence = float(np.max(pred))

        results.append([
            img_name,
            true_class,
            pred_class,
            round(confidence, 3)
        ])

df = pd.DataFrame(results, columns=[
    "image",
    "true_class",
    "predicted_class",
    "confidence"
])

df.head(20)
```

```python
import tensorflow as tf
import numpy as np
import os
from tensorflow.keras.preprocessing import image
import pandas as pd

# Load model
model = tf.keras.models.load_model("warp_waste_classifier.h5")

IMG_SIZE = 224

# FINAL LABELS (MATCH TRAINING)
labels = ['plastic', 'metal', 'paper']
num_classes = model.output_shape[-1]

results = []

base_dir = "Warp-C/val"

for true_class in labels:   # ONLY iterate over trained classes
    class_dir = os.path.join(base_dir, true_class)
    if not os.path.isdir(class_dir):
        continue

    for img_name in os.listdir(class_dir):
        img_path = os.path.join(class_dir, img_name)

        img = image.load_img(img_path, target_size=(IMG_SIZE, IMG_SIZE))
        img_array = image.img_to_array(img) / 255.0
        img_array = np.expand_dims(img_array, axis=0)

        pred = model.predict(img_array, verbose=0)[0]

        # SAFETY CHECK
        if len(pred) != num_classes:
            continue

        pred_index = int(np.argmax(pred))
        pred_class = labels[pred_index]
        confidence = float(pred[pred_index])

        results.append([
            img_name,
            true_class,
            pred_class,
            round(confidence, 3)
        ])

df = pd.DataFrame(results, columns=[
    "image", "true_class", "predicted_class", "confidence"
])

df.head(20)
```

```python
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator

# PARAMETERS (LOCK THESE)
IMG_SIZE = 224
BATCH_SIZE = 32
EPOCHS = 15

train_dir = "Warp-C/train"
val_dir = "Warp-C/val"

# DATA GENERATORS
train_gen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=20,
    zoom_range=0.2,
    horizontal_flip=True
)

val_gen = ImageDataGenerator(rescale=1./255)

train_data = train_gen.flow_from_directory(
    train_dir,
    target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    class_mode='categorical'
)

val_data = val_gen.flow_from_directory(
    val_dir,
    target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    class_mode='categorical'
)

print("CLASS INDICES:", train_data.class_indices)

# MODEL
base_model = tf.keras.applications.MobileNetV2(
    input_shape=(IMG_SIZE, IMG_SIZE, 3),
    include_top=False,
    weights="imagenet"
)

base_model.trainable = False

model = tf.keras.Sequential([
    base_model,
    tf.keras.layers.GlobalAveragePooling2D(),
    tf.keras.layers.Dense(128, activation='relu'),
    tf.keras.layers.Dropout(0.3),
    tf.keras.layers.Dense(3, activation='softmax')  # 🔴 EXACTLY 3
])

model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

history = model.fit(
    train_data,
    validation_data=val_data,
    epochs=EPOCHS
)
```

```python
import tensorflow as tf
import numpy as np
import os
import matplotlib.pyplot as plt
from tensorflow.keras.preprocessing import image

# Load model
model = tf.keras.models.load_model("warp_waste_classifier_3class.h5")

IMG_SIZE = 224
labels = ['metal', 'paper', 'plastic']  # FINAL ORDER

base_dir = "Warp-C/val"

# Limit how many images to display per class (change if needed)
MAX_IMAGES_PER_CLASS = 10

for true_class in labels:
    class_dir = os.path.join(base_dir, true_class)
    print(f"\n🔹 Testing class: {true_class.upper()}\n")

    images = os.listdir(class_dir)[:MAX_IMAGES_PER_CLASS]

    for img_name in images:
        img_path = os.path.join(class_dir, img_name)

        # Load image
        img = image.load_img(img_path, target_size=(IMG_SIZE, IMG_SIZE))
        img_array = image.img_to_array(img) / 255.0
        img_array = np.expand_dims(img_array, axis=0)

        # Predict
        pred = model.predict(img_array, verbose=0)[0]
        pred_index = int(np.argmax(pred))
        pred_class = labels[pred_index]
        confidence = float(pred[pred_index])

        # Display image
        plt.imshow(img)
        plt.axis('off')
        plt.title(
            f"True: {true_class} | Predicted: {pred_class} | Confidence: {confidence:.2f}"
        )
        plt.show()
```
